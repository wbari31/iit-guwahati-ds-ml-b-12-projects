```python
import pandas as pd
import matplotlib.pyplot as plt
import plotly.express as px
import seaborn as sns
import warnings
warnings.filterwarnings('ignore')
```

Data Preparation

```python
# 1. Load the student marks and attendance data from the provided Excel files into two separate Pandas DataFrames.
df_marks = pd.read_excel("Student_performance_and_attendance_data.xlsx", sheet_name="Marks")
display(df_marks)
df_attendance = pd.read_excel("Student_performance_and_attendance_data.xlsx", sheet_name="Attendance")
display(df_attendance)
```

```python
#2. Merge the two DataFrames on the Name column to create a single DataFrame.
df_merged = pd.merge(df_marks, df_attendance, on="Name", how="inner")
display(df_merged)
```

```python
# 3. Replace attendance values ('Y' or 'N') with numeric values (1 for 'Y' and 0 for 'N').
df_merged = df_merged.replace({
    "Attendance Day 1": {"Y": 1, "N": 0},
    "Attendance Day 2": {"Y": 1, "N": 0},
    "Attendance Day 3": {"Y": 1, "N": 0},
    "Attendance Day 4": {"Y": 1, "N": 0},
    "Attendance Day 5": {"Y": 1, "N": 0}
})
display(df_merged)
```

```python
#4.	Deal with the missing values.
for column_name in ['Mini Test 1', 'Mini Test 2', 'Live Test', 'Assignment']:
	mean_val= df_merged[(df_merged[column_name].notnull()) & (df_merged[column_name]!=' ')][column_name].mean()
	df_merged[column_name] = df_merged[column_name].replace(" ",mean_val)
	df_merged[column_name] = df_merged[column_name].fillna(mean_val)
```

```python
#5.	Do the column cleaning like name column is the combination of upper and lower case.
df_merged['Name']=df_merged["Name"].str.title()
```

Data Transformation

```python
#o	Total marks for each student.
df_merged['Total Marks']=df_merged['Mini Test 1'] + df_merged['Mini Test 2'] + df_merged['Live Test'] + df_merged['Assignment']
```

```python
#o	Percentage marks for each student.
df_merged['Total Marks Percent']=df_merged['Total Marks']*100/160
```

```python
#o	Attendance percentage for each student based on attendance columns.
df_merged['Attendance Percent']=(df_merged['Attendance Day 1'] + df_merged['Attendance Day 2']+df_merged['Attendance Day 3']+df_merged['Attendance Day 4']+df_merged['Attendance Day 5']) * 20
```

```python
#o	Calculate the weighted percentage – Attendance(40%), Mini test01(10%), Mini test02(10%), Live test(20%), Assignment (20%).
df_merged['weighted percent']=df_merged['Attendance Percent']*0.4 + df_merged['Mini Test 1']*1 + df_merged['Mini Test 2']*1 + \
(df_merged['Live Test']*100/120)*0.2 + (df_merged['Assignment']*100/20)*0.2
df_merged
```

```python
df_merged['weighted percent'].max()
```

```python
#2.	Classify each student's performance into categories:
#o	"Excellent" for percentages ≥ 85.
#o	"Good" for percentages between 71 and 84.
#o	"Average" for percentages between 50 and 70.
#o	"Needs Improvement" for percentages < 50.

df_merged['Performance']=pd.cut(df_merged['weighted percent'], [0,50,71,85,100], right=False, labels=['Needs Improvement','Average','Good','Excellent'],\
       include_lowest=True, duplicates='raise')
```

Analysis

Visualization

```python
#1.	Create a bar chart displaying weighted percentages for top 5 students.
top5_students = df_merged.nlargest(5, "weighted percent")
fig = px.bar(top5_students, x="Name", y="weighted percent", title="Top 5 Students")
fig.show()
```

```python
#2.	Create a pie chart showing the distribution of students across the four performance categories.
performance_counts=df_merged['Performance'].value_counts()
performance_counts
display(performance_counts)
plt.figure(figsize=(6,6))
plt.pie(performance_counts)
plt.title("Performance Distribution")
plt.show()
```

```python
# 4.	Create a chart to show the students where attendance is less than 50%.
colors = ["red" if  att< 50 else "green" for att in df_merged['Attendance Percent']]
plt.figure(figsize=(20,10))
plt.xticks(rotation=90)
plt.bar(df_merged['Name'],df_merged['Attendance Percent'], color=colors)
plt.axhline(50, color="black", linestyle="--", linewidth=1, label="50% Attendance Threshold")
plt.ylabel("Attendance Percentage")
plt.title("Student Attendance")
plt.legend()
plt.show()
```
